## 3. Success metric

*### 3. Success Metric

* **Primary Metric:** Precision & F1-Score (target threshold: $\ge 0.80$)
* **Secondary Metric:** ROC-AUC Score

#### Rationale:
In a content recommendation and performance prediction pipeline, **Precision** measures how many predicted "High Performers" actually achieve high traffic post-publication. High precision prevents investing resources into publishing content that fails to perform.

## 5. Why ML beats a fixed rule here

*### 5. Why ML Beats a Fixed Rule Here

#### Limitations of Fixed Rules:
A fixed, heuristic rule—such as requiring `word_count > 1500` or `keyword_density > 2%`—is overly rigid and fails to capture the complex dynamics of search performance and content quality:
1. **Non-Linear Relationships & Interactions:** Quality and search ranking depend on non-linear interactions between readability, intent match, keyword placement, and topic depth. A long article is not automatically high-performing if it lacks relevance or readability.
2. **Search Engine Dynamics & Drift:** Search algorithms continuously update, and user behavior shifts over time. A static rule cannot adapt to shifting ranking factors or changing market trends without constant manual rewriting.
3. **Threshold Fragility:** Fixed thresholds (e.g., exactly 1,500 words) create arbitrary pass/fail boundaries that misclassify high-value concise content or promote low-value wordy content.

#### Why Machine Learning Solves This:
* **Learns High-Dimensional Patterns:** ML models automatically discover complex feature interactions (e.g., combining readability, semantic depth, and structure) without human bias.
* **Continuous Adaptation:** As search data and user engagement metrics update, the model can be retrained on new targets, continuously refining its predictions.
* **Probabilistic Outputs:** Instead of a rigid yes/no rule, ML outputs calibrated probabilities that allow teams to set custom confidence thresholds based on business risk.*
